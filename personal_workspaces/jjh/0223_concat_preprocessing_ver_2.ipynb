```python
type(EDA_columns['t_dat'])
EDA_columns['transaction_dat'] = pd.to_datetime(EDA_columns['t_dat'], format="%Y-%m-%d")
```

```python
# 평균 구매 간격
EDA_columns = EDA_columns.sort_values(['customer_id','transaction_dat'])
EDA_columns['prev_date'] = EDA_columns.groupby('customer_id')['transaction_dat'].shift(1)
EDA_columns['gap'] = (EDA_columns['transaction_dat'] - EDA_columns['prev_date']).dt.days
EDA_columns = EDA_columns.fillna(0) # 시작날은 시차 0 
```

```python
EDA_columns['t_dat'] = EDA_columns['t_dat'].astype(str)
```

# 1. 전체 고객 이탈 label : API 적용-> API가 90일보다 크면 이탈 (90일) + 1번 구매 후 없음

```python
EDA_columns['t_dat'] = pd.to_datetime(EDA_columns['t_dat'], format="%Y-%m-%d")

# 1. 기준 날짜 설정 및 LPL 계산
last_date = EDA_columns['t_dat'].max() 

# 고객별 마지막 구매일 확인 및 LPL(경과일) 계산
lpl_df = EDA_columns.groupby('customer_id')['t_dat'].max().reset_index()
lpl_df['LPL'] = (last_date - lpl_df['t_dat']).dt.days

# 2. 거래 특성 요약
API_count = EDA_columns.groupby('customer_id').agg({
    't_dat': 'nunique', # 중복 없는 거래일 수
    'gap': 'sum'        # 날짜 간격 합계
}).reset_index()

# 3. 데이터 병합
API_count = API_count.merge(lpl_df[['customer_id', 'LPL']], on='customer_id', how='left')

# 4. API ratio 계산 (거래 1회인 경우 나누기 에러 방지용 0 처리)
API_count['API ratio'] = np.where(API_count['t_dat'] > 1, 
                                  API_count['gap'] / (API_count['t_dat'] - 1), 
                                  0)

# ---------------------------------------------------------
# 5. 핵심 라벨링 (신규 고객 이탈 vs 기존 고객 이탈)
# ---------------------------------------------------------

# (1) 신규 고객 여부 (거래 횟수가 정확히 1회)
API_count['only_first_tran'] = np.where(API_count['t_dat'] == 1, 1, 0)

# (2) 신규 고객 이탈 라벨 (거래가 1회인 고객은 작성하신 논리대로 무조건 이탈로 표시)
API_count['new_customer_churn'] = np.where(API_count['only_first_tran'] == 1, 1, 0)

# (3) 전체 고객 통합 이탈 라벨 (total_churn)
# 논리: [거래가 1회뿐인 신규 고객] 이거나 [기존 고객 중 주기를 넘긴 사람]
API_count['total_churn'] = np.where(
    (API_count['new_customer_churn'] == 1) | (API_count['LPL'] > API_count['API ratio']), 
    1, 
    0
)

# 6. 결과 확인
print(f"신규 고객(거래 1회) 수: {API_count['new_customer_churn'].sum()}명")
print(f"전체 이탈자(1) 비율:\n{API_count['total_churn'].value_counts(normalize=True)}")
```

```python
# 전체 고객 이탈 라벨링 데이터를 EDA df에 merge (0:미이탈 / 1:이탈)
API_merge = API_count[['customer_id','API ratio','total_churn','new_customer_churn']]
API_churn_label_merge = EDA_columns.merge(API_merge,on=['customer_id'],how='left')

# # 신규 고객 이탈 라벨링 데이터를 EDA df에 merge (0:미이탈 / 1:이탈)
# first_customer_merge = churn_label[['customer_id','신규고객이탈']]
# API_churn_label_merge = API_churn_label_merge.merge(first_customer_merge,on=['customer_id'],how='left' )
# API_churn_label_merge

```
