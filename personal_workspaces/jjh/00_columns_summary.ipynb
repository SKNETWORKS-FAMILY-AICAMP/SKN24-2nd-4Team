| perceived_colour_master_id   | 상품     | 색상 계열 식별 코드           | integer      |
| perceived_colour_master_name | 상품     | 색상 계열 이름              | categorical  |
| department_no                | 상품     | 부서 식별 코드              | integer      |
| department_name              | 상품     | 부서 이름                 | categorical  |
| index_code                   | 상품     | 상품 인덱스 코드             | string       |
| index_name                   | 상품     | 주요 상품 카테고리            | categorical  |
