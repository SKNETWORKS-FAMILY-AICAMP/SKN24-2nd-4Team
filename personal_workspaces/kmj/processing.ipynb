# 위 모든 데이터 결측치 제거 확인

### join 하는 절차

- final_customer_df
- final_transction_df
- final_articles_df

- **customer_id**: 고객을 구분하는 고유 ID (조인 키)
- **FN**: 패션 뉴스/마케팅 정보 수신(또는 관심/동의) 여부
- **Active**: 최근 활동(구매/방문 등) 여부
- **club_member_status**: 클럽/멤버십 가입 상태 
- **fashion_news_frequency**: 패션 뉴스/마케팅 메시지 수신 빈도 

```python
final_customer_df.head(15)
```

```python
final_transction_df.head()
```

```python
final_transction_df['t_dat'] = pd.to_datetime(final_transction_df['t_dat'])
```

```python
cust_feat = final_transction_df.groupby('customer_id').agg(
    n_tx=('t_dat', 'size'),             # 총 구매 횟수(거래 건수)
    last_purchase=('t_dat', 'max'),     # 마지막 구매일
    first_purchase=('t_dat', 'min'),    # 첫 구매일
    n_items=('article_id', 'nunique')   # 구매한 서로 다른 상품 수
).reset_index()
```

```python
customer_merged_df = final_transction_df.merge(
    cust_feat, 
    on='customer_id',
    how='left'
)
```

```python
# gpt...
customer_merged_df['n_tx'] = customer_merged_df['n_tx'].fillna(0)
customer_merged_df['n_items'] = customer_merged_df['n_items'].fillna(0)
```
